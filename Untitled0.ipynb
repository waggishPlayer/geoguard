```python
#!/usr/bin/env python3
# src/sim/sensor_stream.py
"""
SensorStream — immediate-swap behavior: simulated stops as soon as manual is selected.

Key points:
- background_tick checks mode and calls the appropriate internal updater directly.
- set_mode('manual') clears simulated events and zeroes velocities.
- manual updates use only sanitized manual inputs (no injected randomness/events).
"""

import time, random, math, threading
from datetime import datetime

class SensorStream:
    def __init__(self, n_sensors=10, site_id="siteA"):
        self.n_sensors = n_sensors
        self.site_id = site_id
        self.sensors = self._initialize_sensors()
        self.start_time = time.time()
        self.last_update = self.start_time

        # inertia params
        self.k_temp = 0.8; self.c_temp = 0.9
        self.k_pore = 1.0; self.c_pore = 1.2
        self.k_disp = 1.2; self.c_disp = 1.5
        self.k_vib = 1.0; self.c_vib = 1.0
        self.k_tilt = 0.6; self.c_tilt = 1.0

        # safety limits
        self.max_acc = {'temp_c': 5.0, 'pore_kpa': 2.0, 'disp_mm': 5.0, 'vibration_g': 0.05, 'tilt_deg': 0.5}
        self.max_vel = {'temp_c': 2.0, 'pore_kpa': 1.0, 'disp_mm': 10.0, 'vibration_g': 0.2, 'tilt_deg': 1.0}
        self.max_disp_delta = 5.0

        self.cycle_amplitude = {'temp': 5.0, 'pore': 2.0, 'disp': 0.02}
        self.active_events = {'rain': {'active': False, 'intensity': 0.0, 'start_time': 0},
                              'tremor': {'active': False, 'magnitude': 0.0, 'start_time': 0, 'decay': 0.0}}

        self.mode = 'simulated'
        self.manual_conditions = None

        random.seed(42)

    def _initialize_sensors(self):
        sensors = []
        for i in range(self.n_sensors):
            s_type = random.choice(["displacement", "piezometer", "vibration", "tilt"])
            baseline_disp = random.uniform(0.0, 0.5)
            baseline_pore = random.uniform(5.0, 15.0)
            sensors.append({
                "id": f"S{i+1:02d}",
                "type": s_type,
                "lat": 11.102222 + random.uniform(-0.001, 0.001),
                "lon": 79.156389 + random.uniform(-0.001, 0.001),
                "state": {"disp_mm": baseline_disp, "pore_kpa": baseline_pore, "vibration_g": 0.001,
                          "tilt_deg": random.uniform(0.0, 0.05), "temp_c": 25.0},
                "vel": {"disp_mm": 0.0, "pore_kpa": 0.0, "vibration_g": 0.0, "tilt_deg": 0.0, "temp_c": 0.0},
                "baseline": {"disp_mm": baseline_disp, "pore_kpa": baseline_pore}
            })
        return sensors

    # -----------------------
    # Mode control (immediate)
    # -----------------------
    def set_mode(self, mode):
        if mode not in ('simulated', 'manual', 'off'):
            raise ValueError("mode must be 'simulated', 'manual', or 'off'")
        if mode == 'manual':
            # Clear simulated events and reset velocities immediately
            self.stop_simulation()
            self._reset_velocities()
        self.mode = mode

    def _reset_velocities(self):
        for s in self.sensors:
            for k in s['vel']:
                s['vel'][k] = 0.0

    def set_manual_conditions(self, conditions):
        self.manual_conditions = conditions.copy()

    def apply_preset(self, name):
        presets = {
            'clear': {"wind_speed": 10.0, "sun": 0.9, "rain_mm": 0.0, "storm": 0.0, "earthquake": 0.0, "soil_moisture": 0.3},
            'light_rain': {"wind_speed": 10.0, "sun": 0.3, "rain_mm": 5.0, "storm": 0.0, "earthquake": 0.0, "soil_moisture": 0.5},
            'heavy_rain': {"wind_speed": 15.0, "sun": 0.1, "rain_mm": 25.0, "storm": 0.2, "earthquake": 0.0, "soil_moisture": 0.8},
            'storm': {"wind_speed": 45.0, "sun": 0.1, "rain_mm": 40.0, "storm": 0.8, "earthquake": 0.0, "soil_moisture": 0.9},
            'quake': {"wind_speed": 5.0, "sun": 0.5, "rain_mm": 0.0, "storm": 0.0, "earthquake": 0.8, "soil_moisture": 0.6}
        }
        if name not in presets:
            raise ValueError("Unknown preset")
        cfg = presets[name].copy()
        self.set_manual_conditions(cfg)
        self.set_mode('manual')
        return cfg

    def stop_simulation(self):
        self.active_events['rain'] = {'active': False, 'intensity': 0.0, 'start_time': 0}
        self.active_events['tremor'] = {'active': False, 'magnitude': 0.0, 'start_time': 0, 'decay': 0.0}

    # -----------------------
    # Physics helpers
    # -----------------------
    def _apply_daily_cycle(self, timestamp):
        dt = datetime.fromtimestamp(timestamp)
        day_progress = (dt.hour + dt.minute/60.0) / 24.0
        return math.sin(2 * math.pi * (day_progress - 0.25))

    def _clamp(self, x, lo, hi): return max(lo, min(hi, x))

    def _integrate_inertia(self, s, key, target, k, c, dt):
        pos = s['state'][key]; vel = s['vel'][key]
        raw_acc = k * (target - pos) - c * vel

        key_map = {'temp_c': 'temp_c', 'pore_kpa': 'pore_kpa', 'disp_mm': 'disp_mm',
                   'vibration_g': 'vibration_g', 'tilt_deg': 'tilt_deg'}
        km = key_map.get(key)
        if km is not None:
            maxa = self.max_acc.get(km, None)
            if maxa is not None:
                raw_acc = self._clamp(raw_acc, -maxa, maxa)

        vel_new = vel + raw_acc * dt
        if km is not None:
            maxv = self.max_vel.get(km, None)
            if maxv is not None:
                vel_new = self._clamp(vel_new, -maxv, maxv)

        pos_new = pos + vel_new * dt
        s['vel'][key] = vel_new
        s['state'][key] = pos_new

    # -----------------------
    # Internal updaters (called directly by background thread)
    # -----------------------
    def _update_simulated(self, dt):
        now = time.time(); cycle = self._apply_daily_cycle(now)
        for s in self.sensors:
            baseline_disp = s['baseline']['disp_mm']; baseline_pore = s['baseline']['pore_kpa']
            target_temp = 25.0 + self.cycle_amplitude['temp'] * cycle + random.uniform(-0.3, 0.3)
            target_pore = baseline_pore + (-cycle * self.cycle_amplitude['pore']) * 0.2 + random.uniform(-0.2, 0.2)
            target_disp = baseline_disp + self.cycle_amplitude['disp'] * cycle + random.uniform(-0.002, 0.002)

            if self.active_events['rain']['active']:
                intensity = self.active_events['rain']['intensity']
                target_pore += intensity * 0.25 * (0.8 + random.random() * 0.4)
                if s['type'] == 'displacement' and random.random() < 0.35:
                    target_disp += intensity * 0.01 * (0.8 + random.random() * 0.6)

            if self.active_events['tremor']['active']:
                mag = self.active_events['tremor']['magnitude']; decay = self.active_events['tremor']['decay']
                elapsed = now - self.active_events['tremor']['start_time']
                current_vib = mag * math.exp(-decay * elapsed)
                if current_vib < 0.0008:
                    self.active_events['tremor']['active'] = False; current_vib = 0.0008
                target_vib = current_vib + random.uniform(0.0, 0.0015)
                if current_vib > 0.05:
                    target_disp += current_vib * 0.2
                    target_tilt = s['state']['tilt_deg'] + current_vib * 0.08
                else:
                    target_tilt = s['state']['tilt_deg'] + random.uniform(-0.002, 0.002)
            else:
                target_vib = 0.0008 + random.uniform(0.0, 0.0012)
                target_tilt = s['state']['tilt_deg'] + random.uniform(-0.001, 0.001)

            self._integrate_inertia(s, 'temp_c', target_temp, self.k_temp, self.c_temp, dt)
            self._integrate_inertia(s, 'pore_kpa', target_pore, self.k_pore, self.c_pore, dt)
            self._integrate_inertia(s, 'disp_mm', target_disp, self.k_disp, self.c_disp, dt)
            self._integrate_inertia(s, 'vibration_g', target_vib, self.k_vib, self.c_vib, dt)
            self._integrate_inertia(s, 'tilt_deg', target_tilt, self.k_tilt, self.c_tilt, dt)

            s['state']['disp_mm'] = max(0.0, s['state']['disp_mm'])
            s['state']['pore_kpa'] = max(0.0, s['state']['pore_kpa'])
            s['state']['vibration_g'] = max(0.0, s['state']['vibration_g'])

    def _update_manual(self, dt, conditions):
        wind = float(conditions.get("wind_speed", 0.0))
        sun = float(conditions.get("sun", 0.5))
        rain_mm = float(conditions.get("rain_mm", 0.0))
        storm = float(conditions.get("storm", 0.0))
        eq = float(conditions.get("earthquake", 0.0))
        soil = float(conditions.get("soil_moisture", 0.5))

        wind_norm = max(0.0, min(wind / 60.0, 1.0))
        rain_norm = max(0.0, min(rain_mm / 50.0, 1.0))
        sun_norm = max(0.0, min(sun, 1.0))
        storm_norm = max(0.0, min(storm, 1.0))
        eq_norm = max(0.0, min(eq, 1.0))
        soil_norm = max(0.0, min(soil, 1.0))

        for s in self.sensors:
            baseline_disp = s['baseline']['disp_mm']; baseline_pore = s['baseline']['pore_kpa']
            target_temp = 15.0 + 12.0 * sun_norm - 3.0 * rain_norm
            target_pore = baseline_pore + (rain_norm * 6.0) + (soil_norm - 0.5) * 4.0 + storm_norm * 3.0
            disp_offset = rain_norm * 0.07 + (soil_norm - 0.5) * 0.12 + storm_norm * 0.2 + eq_norm * 0.25 + wind_norm * 0.03
            if s['type'] == 'displacement':
                disp_offset *= 1.15
            target_disp = baseline_disp + disp_offset
            target_disp = self._clamp(target_disp, baseline_disp - self.max_disp_delta, baseline_disp + self.max_disp_delta)
            target_vib = 0.0008 + eq_norm * 0.04 + storm_norm * 0.015 + wind_norm * 0.002
            target_tilt = s['state']['tilt_deg'] + (storm_norm * 0.03 + eq_norm * 0.06 + wind_norm * 0.01)

            self._integrate_inertia(s, 'temp_c', target_temp, self.k_temp * 1.1, self.c_temp * 1.0, dt)
            self._integrate_inertia(s, 'pore_kpa', target_pore, self.k_pore * 1.1, self.c_pore * 1.1, dt)
            self._integrate_inertia(s, 'disp_mm', target_disp, self.k_disp * 1.1, self.c_disp * 1.2, dt)
            self._integrate_inertia(s, 'vibration_g', target_vib, self.k_vib, self.c_vib, dt)
            self._integrate_inertia(s, 'tilt_deg', target_tilt, self.k_tilt, self.c_tilt, dt)

            s['state']['disp_mm'] = max(0.0, s['state']['disp_mm'])
            s['state']['pore_kpa'] = max(0.0, s['state']['pore_kpa'])
            s['state']['vibration_g'] = max(0.0, s['state']['vibration_g'])

    # -----------------------
    # Events (simulated)
    # -----------------------
    def trigger_rain(self, intensity=1.0):
        self.active_events['rain']['active'] = True
        self.active_events['rain']['intensity'] = max(0.0, intensity)
        self.active_events['rain']['start_time'] = time.time()
        self.mode = 'simulated'

    def stop_rain(self):
        self.active_events['rain']['active'] = False
        self.active_events['rain']['intensity'] = 0.0

    def trigger_tremor(self, magnitude=0.5, decay=0.5):
        self.active_events['tremor']['active'] = True
        self.active_events['tremor']['magnitude'] = max(0.0, magnitude)
        self.active_events['tremor']['start_time'] = time.time()
        self.active_events['tremor']['decay'] = decay
        self.mode = 'simulated'

    # -----------------------
    # Single-tick advance helper (background calls this)
    # -----------------------
    def advance_one_tick(self, dt):
        """
        Advance internal state by dt according to current mode.
        This method is used by the background thread so mode switching takes effect immediately.
        """
        if self.mode == 'simulated':
            self._update_simulated(dt)
        elif self.mode == 'manual':
            if not self.manual_conditions:
                # if manual selected but no values yet, do nothing (keep last state)
                return
            self._update_manual(dt, self.manual_conditions)
        elif self.mode == 'off':
            return
        else:
            raise RuntimeError("unknown mode")

    def get_latest_readings(self):
        # keep last_update bookkeeping consistent (so dt computed correctly)
        now = time.time()
        dt = max(now - self.last_update, 1e-3)
        dt = min(dt, 2.0)
        # advance according to current mode immediately (same as background does)
        self.advance_one_tick(dt)
        self.last_update = now

        timestamp = datetime.now().isoformat()
        readings = []
        for s in self.sensors:
            readings.append({
                "sensor_id": s['id'],
                "type": s['type'],
                "timestamp": timestamp,
                "values": {
                    "disp_mm": round(s['state']['disp_mm'], 4),
                    "pore_kpa": round(s['state']['pore_kpa'], 2),
                    "vibration_g": round(s['state']['vibration_g'], 5),
                    "tilt_deg": round(s['state']['tilt_deg'], 4),
                    "temp_c": round(s['state']['temp_c'], 1)
                },
                "location": {"lat": s['lat'], "lon": s['lon']}
            })
        return readings

# -------------------------
# sanitization helper (unchanged)
# -------------------------
def sanitize_manual_inputs(inputs):
    ranges = {"wind_speed": (0.0, 100.0), "sun": (0.0, 1.0), "rain_mm": (0.0, 200.0),
              "storm": (0.0, 1.0), "earthquake": (0.0, 1.0), "soil_moisture": (0.0, 1.0)}
    sanitized = {}; warnings = []
    defaults = {"wind_speed": 10.0, "sun": 0.6, "rain_mm": 0.0, "storm": 0.0, "earthquake": 0.0, "soil_moisture": 0.4}
    for k, (lo, hi) in ranges.items():
        raw = inputs.get(k, None)
        if raw is None:
            sanitized[k] = defaults[k]; continue
        try:
            v = float(raw)
        except (TypeError, ValueError):
            v = defaults[k]; warnings.append(f"{k} invalid; defaulting to {v}")
        v_clamped = max(lo, min(hi, v)); sanitized[k] = v_clamped
        if v != v_clamped:
            warnings.append(f"{k}={v} clamped to {v_clamped} (allowed range {lo}-{hi})")
    return sanitized, warnings

# -------------------------
# CLI + background threads (background uses advance_one_tick so mode switch is immediate)
# -------------------------
if __name__ == "__main__":
    stream = SensorStream(n_sensors=6)

    # initial choice
    initial = ""
    while initial not in ("simulated", "manual"):
        initial = input("Start in which mode? [simulated/manual] (default: simulated): ").strip().lower() or "simulated"

    if initial == "manual":
        print("Provide manual parameters (press Enter to accept defaults).")
        defaults = {"wind_speed": 10.0, "sun": 0.6, "rain_mm": 0.0, "storm": 0.0, "earthquake": 0.0, "soil_moisture": 0.4}
        user_inputs = {}
        for k, v in defaults.items():
            raw = input(f"  {k} [{v}]: ").strip()
            user_inputs[k] = raw if raw != "" else v
        sanitized, warns = sanitize_manual_inputs(user_inputs)
        if warns:
            print("Note:"); [print(" -", w) for w in warns]
        stream.set_manual_conditions(sanitized)
        stream.set_mode('manual')
        print("Starting in manual mode with:", sanitized)
    else:
        stream.set_mode('simulated')
        print("Starting in simulated mode (live).")

    latest_lock = threading.Lock()
    latest_snapshot = {"readings": stream.get_latest_readings(), "mode": stream.mode}

    def background_tick(interval=1.0):
        while True:
            try:
                now = time.time()
                # compute dt relative to last_update to keep integrator stable
                dt = max(now - stream.last_update, 1e-3)
                dt = min(dt, 2.0)
                # advance one tick according to current mode (this obeys manual immediately)
                stream.advance_one_tick(dt)
                stream.last_update = now
                with latest_lock:
                    latest_snapshot["readings"] = stream.get_latest_readings()
                    latest_snapshot["mode"] = stream.mode
            except Exception as e:
                print("Background tick error:", e)
            time.sleep(interval)

    def live_printer(interval=1.0):
        while True:
            with latest_lock:
                snap = dict(latest_snapshot)
            first = snap["readings"][0]
            print(f"[mode={snap['mode']}] {first['timestamp']} {first['values']}")
            time.sleep(interval)

    ticker = threading.Thread(target=background_tick, args=(1.0,), daemon=True)
    printer = threading.Thread(target=live_printer, args=(1.0,), daemon=True)
    ticker.start(); printer.start()

    print("Commands: 'show'/'Enter' print latest, 'm' switch to manual, 'p' preset, 'r' trigger rain, 's' stop rain, 't' tremor, 'auto' simulated, 'q' quit")

    try:
        while True:
            cmd = input("cmd> ").strip().lower()
            if cmd in ("", "show"):
                with latest_lock:
                    snap = dict(latest_snapshot)
                first = snap["readings"][0]
                print(f"[mode={snap['mode']}] {first['timestamp']} {first['values']}")

            elif cmd == "m":
                print("Switching to manual — enter manual parameters.")
                defaults = {"wind_speed": 10.0, "sun": 0.6, "rain_mm": 0.0, "storm": 0.0, "earthquake": 0.0, "soil_moisture": 0.4}
                user_inputs = {}
                for k, v in defaults.items():
                    raw = input(f"  {k} [{v}]: ").strip()
                    user_inputs[k] = raw if raw != "" else v
                sanitized, warns = sanitize_manual_inputs(user_inputs)
                if warns:
                    print("Note:"); [print(" -", w) for w in warns]
                stream.set_manual_conditions(sanitized)
                stream.set_mode('manual')  # clears events and resets velocities immediately
                with latest_lock:
                    latest_snapshot["readings"] = stream.get_latest_readings()
                    latest_snapshot["mode"] = stream.mode
                print("Now in manual mode with:", sanitized)

            elif cmd == "p":
                print("Presets: clear, light_rain, heavy_rain, storm, quake")
                name = input("preset> ").strip()
                try:
                    cfg = stream.apply_preset(name)
                    with latest_lock:
                        latest_snapshot["readings"] = stream.get_latest_readings()
                        latest_snapshot["mode"] = stream.mode
                    print("Applied preset and switched to manual mode:", cfg)
                except Exception as e:
                    print("Error:", e)

            elif cmd == "r":
                i = input("rain intensity (0.1-3.0) [1.0]> ").strip() or "1.0"
                try: intensity = float(i)
                except: intensity = 1.0
                stream.trigger_rain(intensity)
                with latest_lock:
                    latest_snapshot["readings"] = stream.get_latest_readings()
                    latest_snapshot["mode"] = stream.mode
                print(f"Triggered rain (intensity={intensity}) and switched to simulated mode.")

            elif cmd == "s":
                stream.stop_rain()
                with latest_lock:
                    latest_snapshot["readings"] = stream.get_latest_readings()
                    latest_snapshot["mode"] = stream.mode
                print("Stopped rain.")

            elif cmd == "t":
                m = input("tremor magnitude g (0.1-1.5) [0.5]> ").strip() or "0.5"
                try: mag = float(m)
                except: mag = 0.5
                stream.trigger_tremor(mag)
                with latest_lock:
                    latest_snapshot["readings"] = stream.get_latest_readings()
                    latest_snapshot["mode"] = stream.mode
                print(f"Triggered tremor (magnitude={mag}) and switched to simulated mode.")

            elif cmd == "auto":
                stream.set_mode('simulated')
                with latest_lock:
                    latest_snapshot["readings"] = stream.get_latest_readings()
                    latest_snapshot["mode"] = stream.mode
                print("Switched to simulated mode.")

            elif cmd == "q":
                print("Exiting."); break

            else:
                print("Unknown command. Type 'show' or press Enter to view latest.")
    except KeyboardInterrupt:
        print("\nDemo stopped.")
```
